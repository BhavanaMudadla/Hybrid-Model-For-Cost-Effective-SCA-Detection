```python

# ================================================================
# SICKLE CELL CLASSIFICATION USING VGG-16
# COMPLETE GOOGLE COLAB IMPLEMENTATION IN ONE CELL
#
# Dataset:
#   negative/            -> Label 0
#   positive_unlabelled/ -> Label 1
#   positive_labelled/   -> Excluded (black rectangle annotations)
#
# Outputs:
#   Trained VGG-16 model
#   Training curves
#   Confusion matrix
#   ROC curve
#   Test metrics
#   Test predictions
# ================================================================

# ----------------------- 1. INSTALL -------------------------------
!pip -q install albumentations opencv-python-headless scikit-learn matplotlib seaborn

# ----------------------- 2. IMPORTS -------------------------------
from google.colab import drive
drive.mount("/content/drive")

import os
import json
import time
import random
import warnings
import numpy as np
import pandas as pd
import cv2
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
import albumentations as A

from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report,
    roc_curve
)

from tensorflow.keras import layers, models, regularizers
from tensorflow.keras.applications import VGG16
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)

warnings.filterwarnings("ignore")

# ----------------------- 3. REPRODUCIBILITY ----------------------
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("TensorFlow version:", tf.__version__)
print("GPU available:", len(tf.config.list_physical_devices("GPU")) > 0)

# ----------------------- 4. CONFIGURATION ------------------------
DATA_DIR = "/content/drive/MyDrive/Dataset_SickleCell"

NEGATIVE_DIR = os.path.join(DATA_DIR, "negative")
POSITIVE_UNLABELED_DIR = os.path.join(DATA_DIR, "positive_unlabelled")
POSITIVE_LABELED_DIR = os.path.join(DATA_DIR, "positive_labelled")

OUTPUT_DIR = os.path.join(DATA_DIR, "VGG16_model_outputs")
os.makedirs(OUTPUT_DIR, exist_ok=True)

IMG_SIZE = 224
BATCH_SIZE = 8
EPOCHS = 30
THRESHOLD = 0.50

# Confirm folders
for folder in [
    NEGATIVE_DIR,
    POSITIVE_UNLABELED_DIR,
    POSITIVE_LABELED_DIR
]:
    print(folder, "->", os.path.isdir(folder))

if not os.path.isdir(NEGATIVE_DIR):
    raise FileNotFoundError(f"Folder not found: {NEGATIVE_DIR}")

if not os.path.isdir(POSITIVE_UNLABELED_DIR):
    raise FileNotFoundError(
        f"Folder not found: {POSITIVE_UNLABELED_DIR}"
    )

print("\nAnnotated positive images will be excluded.")

# ----------------------- 5. LOAD IMAGES ---------------------------
VALID_EXTENSIONS = (
    ".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff"
)

def load_images_from_folder(folder, img_size=224):
    images = []
    paths = []
    failed = []

    for root, _, files in os.walk(folder):
        for filename in sorted(files):
            if not filename.lower().endswith(VALID_EXTENSIONS):
                continue

            path = os.path.join(root, filename)

            try:
                image = cv2.imread(path)

                if image is None:
                    failed.append(path)
                    continue

                image = cv2.cvtColor(
                    image, cv2.COLOR_BGR2RGB
                )

                image = cv2.resize(
                    image,
                    (img_size, img_size),
                    interpolation=cv2.INTER_AREA
                )

                images.append(image)
                paths.append(path)

            except Exception as error:
                print("Error loading:", path, error)
                failed.append(path)

    if len(images) == 0:
        raise ValueError(f"No readable images found in {folder}")

    images = np.asarray(images, dtype=np.uint8)

    print(f"\nFolder: {folder}")
    print("Images loaded:", len(images))
    print("Images failed:", len(failed))

    return images, paths


X_negative, paths_negative = load_images_from_folder(
    NEGATIVE_DIR, IMG_SIZE
)

X_positive, paths_positive = load_images_from_folder(
    POSITIVE_UNLABELED_DIR, IMG_SIZE
)

print("\nTotal negative images:", len(X_negative))
print("Total unannotated positive images:", len(X_positive))

# ----------------------- 6. SPLIT DATASET -------------------------
# 70% training, 15% validation, 15% testing per class.
#
# Important:
# If multiple images belong to the same patient or slide,
# use patient/slide-level splitting instead of image-level splitting.

def split_class(images, paths, seed=42):
    indices = np.arange(len(images))

    train_idx, temp_idx = train_test_split(
        indices,
        test_size=0.30,
        random_state=seed
    )

    val_idx, test_idx = train_test_split(
        temp_idx,
        test_size=0.50,
        random_state=seed
    )

    return {
        "train_X": images[train_idx],
        "train_paths": [paths[i] for i in train_idx],
        "val_X": images[val_idx],
        "val_paths": [paths[i] for i in val_idx],
        "test_X": images[test_idx],
        "test_paths": [paths[i] for i in test_idx]
    }


negative_split = split_class(
    X_negative, paths_negative, SEED
)

positive_split = split_class(
    X_positive, paths_positive, SEED
)

# Training
X_train = np.concatenate([
    negative_split["train_X"],
    positive_split["train_X"]
])

y_train = np.concatenate([
    np.zeros(len(negative_split["train_X"]), dtype=np.int32),
    np.ones(len(positive_split["train_X"]), dtype=np.int32)
])

paths_train = (
    negative_split["train_paths"]
    + positive_split["train_paths"]
)

# Validation
X_val = np.concatenate([
    negative_split["val_X"],
    positive_split["val_X"]
])

y_val = np.concatenate([
    np.zeros(len(negative_split["val_X"]), dtype=np.int32),
    np.ones(len(positive_split["val_X"]), dtype=np.int32)
])

paths_val = (
    negative_split["val_paths"]
    + positive_split["val_paths"]
)

# Testing
X_test = np.concatenate([
    negative_split["test_X"],
    positive_split["test_X"]
])

y_test = np.concatenate([
    np.zeros(len(negative_split["test_X"]), dtype=np.int32),
    np.ones(len(positive_split["test_X"]), dtype=np.int32)
])

paths_test = (
    negative_split["test_paths"]
    + positive_split["test_paths"]
)

# Shuffle training data
rng = np.random.default_rng(SEED)
order = rng.permutation(len(X_train))

X_train = X_train[order]
y_train = y_train[order]
paths_train = [paths_train[i] for i in order]

print("\n========== DATASET SPLIT ==========")

for name, labels in [
    ("Training", y_train),
    ("Validation", y_val),
    ("Test", y_test)
]:
    print(
        f"{name}: {len(labels)} images | "
        f"Negative: {int(np.sum(labels == 0))} | "
        f"Positive: {int(np.sum(labels == 1))}"
    )

if len(np.unique(y_train)) < 2:
    raise ValueError("Training data must contain both classes.")

if len(np.unique(y_val)) < 2:
    raise ValueError(
        "Validation set needs both classes. "
        "Add more images or revise the split."
    )

if len(np.unique(y_test)) < 2:
    raise ValueError(
        "Test set needs both classes. "
        "Add more images or revise the split."
    )

# Save split paths for reproducibility
with open(
    os.path.join(OUTPUT_DIR, "dataset_split_paths.json"), "w"
) as f:
    json.dump({
        "train": paths_train,
        "validation": paths_val,
        "test": paths_test
    }, f, indent=2)

# ----------------------- 7. DISPLAY SAMPLE IMAGES ---------------
fig, axes = plt.subplots(2, 5, figsize=(15, 6))

for i in range(5):
    axes[0, i].imshow(X_negative[i % len(X_negative)])
    axes[0, i].set_title("Negative")
    axes[0, i].axis("off")

    axes[1, i].imshow(X_positive[i % len(X_positive)])
    axes[1, i].set_title("Positive")
    axes[1, i].axis("off")

plt.tight_layout()
plt.show()

# ----------------------- 8. DATA AUGMENTATION --------------------
# Augmentation is applied to training images only.

augmentation = A.Compose([
    A.Rotate(limit=15, p=0.50),
    A.RandomBrightnessContrast(
        brightness_limit=0.15,
        contrast_limit=0.15,
        p=0.40
    ),
    A.GaussianBlur(blur_limit=(3, 5), p=0.15),
    A.CLAHE(p=0.15),
    A.Perspective(scale=(0.02, 0.05), p=0.15)
])

def augment_numpy(image):
    image = image.astype(np.uint8)
    augmented = augmentation(image=image)
    return augmented["image"].astype(np.uint8)

# VGG16 needs VGG16-specific preprocessing.
def vgg_preprocess(image):
    image = tf.cast(image, tf.float32)
    return tf.keras.applications.vgg16.preprocess_input(image)

def augment_and_preprocess(image, label):
    image = tf.numpy_function(
        func=augment_numpy,
        inp=[image],
        Tout=tf.uint8
    )
    image.set_shape([IMG_SIZE, IMG_SIZE, 3])
    image = vgg_preprocess(image)

    return image, tf.cast(label, tf.float32)

def preprocess_only(image, label):
    image = vgg_preprocess(image)
    return image, tf.cast(label, tf.float32)

AUTOTUNE = tf.data.AUTOTUNE

train_ds = (
    tf.data.Dataset.from_tensor_slices((X_train, y_train))
    .shuffle(
        len(X_train),
        seed=SEED,
        reshuffle_each_iteration=True
    )
    .map(
        augment_and_preprocess,
        num_parallel_calls=AUTOTUNE
    )
    .batch(BATCH_SIZE)
    .prefetch(AUTOTUNE)
)

val_ds = (
    tf.data.Dataset.from_tensor_slices((X_val, y_val))
    .map(preprocess_only, num_parallel_calls=AUTOTUNE)
    .batch(BATCH_SIZE)
    .prefetch(AUTOTUNE)
)

test_ds = (
    tf.data.Dataset.from_tensor_slices((X_test, y_test))
    .map(preprocess_only, num_parallel_calls=AUTOTUNE)
    .batch(BATCH_SIZE)
    .prefetch(AUTOTUNE)
)

print("\nVGG16 data pipelines created.")

# ----------------------- 9. CLASS WEIGHTS ------------------------
classes = np.unique(y_train)

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y_train
)

class_weights = {
    int(cls): float(weight)
    for cls, weight in zip(classes, weights)
}

print("Class weights:", class_weights)

# ----------------------- 10. BUILD VGG16 -------------------------
vgg_base = VGG16(
    weights="imagenet",
    include_top=False,
    input_shape=(IMG_SIZE, IMG_SIZE, 3)
)

# Freeze earlier layers; fine-tune the final convolutional block.
vgg_base.trainable = True

for layer in vgg_base.layers[:-4]:
    layer.trainable = False

inputs = layers.Input(
    shape=(IMG_SIZE, IMG_SIZE, 3),
    name="image_input"
)

x = vgg_base(inputs, training=False)
x = layers.GlobalAveragePooling2D()(x)

x = layers.Dense(
    512,
    activation="relu",
    kernel_regularizer=regularizers.l2(0.001)
)(x)

x = layers.Dropout(0.50)(x)

x = layers.Dense(
    256,
    activation="relu",
    kernel_regularizer=regularizers.l2(0.001)
)(x)

x = layers.Dropout(0.30)(x)

outputs = layers.Dense(
    1,
    activation="sigmoid",
    name="positive_probability"
)(x)

model = models.Model(
    inputs=inputs,
    outputs=outputs,
    name="VGG16_SickleCell"
)

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-4
    ),
    loss=tf.keras.losses.BinaryCrossentropy(),
    metrics=[
        tf.keras.metrics.BinaryAccuracy(name="accuracy"),
        tf.keras.metrics.AUC(name="auc")
    ]
)

model.summary()

# ----------------------- 11. TRAIN VGG16 -------------------------
checkpoint_path = os.path.join(
    OUTPUT_DIR, "best_vgg16.keras"
)

callbacks = [
    ModelCheckpoint(
        filepath=checkpoint_path,
        monitor="val_loss",
        save_best_only=True,
        mode="min",
        verbose=1
    ),
    EarlyStopping(
        monitor="val_loss",
        patience=4,
        restore_best_weights=True,
        verbose=1
    ),
    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.2,
        patience=2,
        min_lr=1e-7,
        verbose=1
    )
]

start_time = time.time()

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS,
    class_weight=class_weights,
    callbacks=callbacks,
    verbose=1
)

training_seconds = time.time() - start_time
training_minutes = training_seconds / 60

print(f"\nTraining time: {training_minutes:.2f} minutes")

# ----------------------- 12. TRAINING CURVES ---------------------
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

axes[0].plot(history.history["loss"], label="Training")
axes[0].plot(history.history["val_loss"], label="Validation")
axes[0].set_title("VGG16 Loss")
axes[0].set_xlabel("Epoch")
axes[0].legend()

axes[1].plot(history.history["accuracy"], label="Training")
axes[1].plot(history.history["val_accuracy"], label="Validation")
axes[1].set_title("VGG16 Accuracy")
axes[1].set_xlabel("Epoch")
axes[1].legend()

axes[2].plot(history.history["auc"], label="Training")
axes[2].plot(history.history["val_auc"], label="Validation")
axes[2].set_title("VGG16 ROC-AUC")
axes[2].set_xlabel("Epoch")
axes[2].legend()

plt.tight_layout()
plt.savefig(
    os.path.join(OUTPUT_DIR, "training_curves.png"),
    dpi=300,
    bbox_inches="tight"
)
plt.show()

# ----------------------- 13. TEST EVALUATION ---------------------
best_model = tf.keras.models.load_model(checkpoint_path)

test_loss, test_accuracy, test_auc = best_model.evaluate(
    test_ds,
    verbose=1
)

y_prob = best_model.predict(test_ds, verbose=1).ravel()
y_pred = (y_prob >= THRESHOLD).astype(np.int32)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)
roc_auc = roc_auc_score(y_test, y_prob)
average_precision = average_precision_score(y_test, y_prob)

metrics = {
    "model": "VGG16",
    "image_size": IMG_SIZE,
    "threshold": THRESHOLD,
    "test_loss": float(test_loss),
    "test_accuracy": float(accuracy),
    "test_precision": float(precision),
    "test_recall": float(recall),
    "test_f1_score": float(f1),
    "test_roc_auc": float(roc_auc),
    "test_average_precision": float(average_precision),
    "training_time_minutes": float(training_minutes),
    "training_images": int(len(X_train)),
    "validation_images": int(len(X_val)),
    "test_images": int(len(X_test)),
    "annotated_positive_folder_used": False
}

print("\n========== VGG16 TEST RESULTS ==========")

for key, value in metrics.items():
    print(f"{key}: {value}")

print("\nClassification report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["Negative", "Positive"],
    zero_division=0
))

# ----------------------- 14. CONFUSION MATRIX --------------------
cm = confusion_matrix(y_test, y_pred, labels=[0, 1])

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Predicted Negative", "Predicted Positive"],
    yticklabels=["Actual Negative", "Actual Positive"]
)
plt.title("VGG16 Test Confusion Matrix")
plt.tight_layout()
plt.savefig(
    os.path.join(OUTPUT_DIR, "confusion_matrix.png"),
    dpi=300,
    bbox_inches="tight"
)
plt.show()

# ----------------------- 15. ROC CURVE ---------------------------
fpr, tpr, roc_thresholds = roc_curve(y_test, y_prob)

plt.figure(figsize=(7, 6))
plt.plot(
    fpr, tpr,
    label=f"VGG16 (AUC = {roc_auc:.3f})"
)
plt.plot([0, 1], [0, 1], "--", label="Chance baseline")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("VGG16 ROC Curve — Test Set")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(
    os.path.join(OUTPUT_DIR, "roc_curve.png"),
    dpi=300,
    bbox_inches="tight"
)
plt.show()

# ----------------------- 16. SAVE MODEL AND RESULTS --------------
final_model_path = os.path.join(
    OUTPUT_DIR, "final_vgg16_sickle_cell.keras"
)
best_model.save(final_model_path)

with open(
    os.path.join(OUTPUT_DIR, "test_metrics.json"), "w"
) as f:
    json.dump(metrics, f, indent=4)

predictions_df = pd.DataFrame({
    "image_path": paths_test,
    "actual_label": y_test,
    "predicted_probability": y_prob,
    "predicted_label": y_pred
})

predictions_df.to_csv(
    os.path.join(OUTPUT_DIR, "test_predictions.csv"),
    index=False
)

print("\n========== SAVED FILES ==========")
for filename in sorted(os.listdir(OUTPUT_DIR)):
    print(os.path.join(OUTPUT_DIR, filename))

# ----------------------- 17. PREDICT A NEW IMAGE -----------------
def predict_sickle_cell_image(image_path, model=best_model):
    image = cv2.imread(image_path)

    if image is None:
        raise ValueError(f"Could not read image: {image_path}")

    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    image = cv2.resize(
        image,
        (IMG_SIZE, IMG_SIZE),
        interpolation=cv2.INTER_AREA
    )

    batch = np.expand_dims(image, axis=0).astype(np.float32)

    batch = tf.keras.applications.vgg16.preprocess_input(batch)

    probability = float(model.predict(batch, verbose=0)[0][0])

    prediction = (
        "Positive" if probability >= THRESHOLD
        else "Negative"
    )

    plt.figure(figsize=(5, 5))
    plt.imshow(image)
    plt.axis("off")
    plt.title(
        f"VGG16 prediction: {prediction}\n"
        f"Positive-class probability: {probability:.4f}"
    )
    plt.show()

    print("Prediction:", prediction)
    print(f"Positive-class probability: {probability:.4f}")

    return prediction, probability


# To predict a new image, uncomment and edit:
# new_image_path = "/content/drive/MyDrive/path_to_new_image.png"
# prediction, probability = predict_sickle_cell_image(new_image_path)

print("\n================================================")
print("VGG16 PIPELINE COMPLETED")
print("Model saved at:", final_model_path)
print("Metrics saved at:", os.path.join(OUTPUT_DIR, "test_metrics.json"))
print("All outputs saved in:", OUTPUT_DIR)
print("================================================")
```
