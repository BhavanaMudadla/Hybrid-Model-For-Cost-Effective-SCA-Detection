```python

# =====================================================================
# DENSENET121 + VGG16 FEATURE-FUSION ENSEMBLE
# Corrected image decoding and preprocessing pipeline
#
# Uses:
#   negative            -> class 0
#   positive_unlabelled -> class 1, configurable subset
#
# Excludes:
#   positive_labelled   -> never loaded
#
# Fixes:
#   1. Validates source images with PIL.
#   2. Converts valid images to standardized RGB PNG files.
#   3. Uses tf.io.decode_png during training.
#   4. Returns ONE image tensor to the ONE-input model.
#   5. Groups exact duplicate image contents before splitting.
#   6. Selects model phase using validation AUC, not test results.
# =====================================================================

import os
import gc
import random
import hashlib
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt

from pathlib import Path
from PIL import Image, ImageFile
from google.colab import drive

from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay,
    roc_curve,
)

# =====================================================================
# 1. CONFIGURATION
# =====================================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

drive.mount("/content/drive")

DATASET_DIR = Path("/content/drive/MyDrive/Dataset_SickleCell")

OUTPUT_DIR = (
    DATASET_DIR / "DenseNet121_VGG16_Unlabelled_Ensemble_Results"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

NEGATIVE_DIR = DATASET_DIR / "negative"
POSITIVE_DIR = DATASET_DIR / "positive_unlabelled"

# Fraction of unique positive-unlabelled images to use.
# 0.80 = 80%; 0.50 = 50%; 1.00 = all.
POSITIVE_FRACTION = 0.80

IMAGE_SIZE = 224
BATCH_SIZE = 8

HEAD_EPOCHS = 10
FINE_TUNE_EPOCHS = 15

TEST_FRACTION = 0.20
VALIDATION_FRACTION = 0.20

IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png", ".bmp",
    ".tif", ".tiff", ".webp", ".gif"
}

AUTOTUNE = tf.data.AUTOTUNE

print("TensorFlow version:", tf.__version__)
print("Dataset:", DATASET_DIR)
print("Output:", OUTPUT_DIR)
print("Positive subset fraction:", POSITIVE_FRACTION)
print("Excluded folder: positive_labelled")

if not 0 < POSITIVE_FRACTION <= 1:
    raise ValueError("POSITIVE_FRACTION must be in (0, 1].")

if not NEGATIVE_DIR.is_dir():
    raise FileNotFoundError(f"Missing folder: {NEGATIVE_DIR}")

if not POSITIVE_DIR.is_dir():
    raise FileNotFoundError(f"Missing folder: {POSITIVE_DIR}")


# =====================================================================
# 2. CONVERT VALID SOURCE IMAGES TO STANDARDIZED RGB PNG
# =====================================================================
# This preprocessing step prevents TensorFlow from having to decode
# the original JPEG, TIFF, GIF, WebP, or potentially malformed files.
#
# Converted PNGs are stored inside the results folder, not the
# original class folders. Only the two intended source folders are read.
#
# The image-content hash is calculated from decoded RGB pixels,
# so exact pixel-identical copies are grouped even if source formats
# differ. This is not a detector for visually similar images.

CONVERTED_DIR = OUTPUT_DIR / "converted_images"

NEGATIVE_CONVERTED_DIR = CONVERTED_DIR / "negative"
POSITIVE_CONVERTED_DIR = CONVERTED_DIR / "positive_unlabelled"

NEGATIVE_CONVERTED_DIR.mkdir(parents=True, exist_ok=True)
POSITIVE_CONVERTED_DIR.mkdir(parents=True, exist_ok=True)


def convert_folder_to_png(source_folder, destination_folder, label):
    records = []
    skipped = 0

    source_files = sorted(
        p for p in source_folder.rglob("*")
        if p.is_file()
        and p.suffix.lower() in IMAGE_EXTENSIONS
    )

    print(f"\nConverting folder: {source_folder.name}")
    print("Source files found:", len(source_files))

    for index, source_path in enumerate(source_files, start=1):
        try:
            # Fully decode the source and normalize it to RGB.
            with Image.open(source_path) as img:
                img.load()
                rgb_image = img.convert("RGB")

                # Hash the exact decoded RGB pixel content and dimensions.
                pixel_hash = hashlib.sha256()
                pixel_hash.update(str(rgb_image.size).encode("utf-8"))
                pixel_hash.update(rgb_image.tobytes())
                image_hash = pixel_hash.hexdigest()

                destination_path = (
                    destination_folder / f"{image_hash}.png"
                )

                # PNG is lossless. Keep the normalized RGB image.
                if not destination_path.exists():
                    rgb_image.save(
                        destination_path,
                        format="PNG",
                        optimize=False
                    )

            records.append({
                "path": str(destination_path),
                "source_path": str(source_path),
                "label": label,
                "hash": image_hash,
            })

        except Exception as exc:
            skipped += 1
            print(
                f"Skipping unreadable file: {source_path.name} "
                f"({type(exc).__name__})"
            )

        if index % 100 == 0:
            print(f"Processed {index}/{len(source_files)} files")

    print("Successfully converted:", len(records))
    print("Skipped unreadable files:", skipped)

    return records


negative_records = convert_folder_to_png(
    NEGATIVE_DIR,
    NEGATIVE_CONVERTED_DIR,
    label=0
)

positive_records = convert_folder_to_png(
    POSITIVE_DIR,
    POSITIVE_CONVERTED_DIR,
    label=1
)

if not negative_records or not positive_records:
    raise ValueError(
        "No usable images remain in one or both required classes."
    )

all_df = pd.DataFrame(negative_records + positive_records)

print("\nConversion summary")
print("Valid image files:", len(all_df))
print("Unique decoded image contents:", all_df["hash"].nunique())


# =====================================================================
# 3. CHECK FOR EXACT DUPLICATES WITH CONFLICTING LABELS
# =====================================================================

hash_label_counts = all_df.groupby("hash")["label"].nunique()

conflicting_hashes = hash_label_counts[
    hash_label_counts > 1
]

if len(conflicting_hashes) > 0:
    conflict_df = all_df[
        all_df["hash"].isin(conflicting_hashes.index)
    ]

    conflict_df.to_csv(
        OUTPUT_DIR / "conflicting_duplicate_labels.csv",
        index=False
    )

    raise ValueError(
        f"Found {len(conflicting_hashes)} pixel-identical images "
        "with conflicting class labels. Review "
        "'conflicting_duplicate_labels.csv' before training."
    )


# =====================================================================
# 4. DEDUPLICATE, THEN SELECT THE POSITIVE SUBSET
# =====================================================================

# Keep one file per exact image hash.
unique_df = (
    all_df.sort_values("path")
    .drop_duplicates(subset="hash", keep="first")
    .reset_index(drop=True)
)

negative_df = unique_df[
    unique_df["label"] == 0
].copy()

positive_df = unique_df[
    unique_df["label"] == 1
].copy()

positive_hashes = positive_df["hash"].to_numpy().copy()

rng = np.random.default_rng(SEED)
rng.shuffle(positive_hashes)

n_positive_keep = max(
    1,
    int(round(len(positive_hashes) * POSITIVE_FRACTION))
)

selected_positive_hashes = set(
    positive_hashes[:n_positive_keep]
)

selected_positive_df = positive_df[
    positive_df["hash"].isin(selected_positive_hashes)
].copy()

# Retain all unique negative images and the selected positive subset.
selected_df = pd.concat(
    [negative_df, selected_positive_df],
    ignore_index=True
)

selected_df = selected_df.sample(
    frac=1,
    random_state=SEED
).reset_index(drop=True)

print("\nSelected data")
print("Negative images:", int((selected_df["label"] == 0).sum()))
print(
    "Positive unlabelled images:",
    int((selected_df["label"] == 1).sum())
)
print("Total unique selected images:", len(selected_df))
print("Positive-labelled images used: 0")

selected_df.to_csv(
    OUTPUT_DIR / "selected_images.csv",
    index=False
)

if selected_df["label"].value_counts().min() < 5:
    raise ValueError(
        "Too few unique images in one class for a reliable stratified split."
    )


# =====================================================================
# 5. STRATIFIED TRAIN / VALIDATION / TEST SPLIT
# =====================================================================

trainval_df, test_df = train_test_split(
    selected_df,
    test_size=TEST_FRACTION,
    random_state=SEED,
    stratify=selected_df["label"]
)

train_df, val_df = train_test_split(
    trainval_df,
    test_size=VALIDATION_FRACTION,
    random_state=SEED,
    stratify=trainval_df["label"]
)

train_df = train_df.reset_index(drop=True)
val_df = val_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

# Confirm that no exact image hash crosses split boundaries.
train_hashes = set(train_df["hash"])
val_hashes = set(val_df["hash"])
test_hashes = set(test_df["hash"])

assert train_hashes.isdisjoint(val_hashes)
assert train_hashes.isdisjoint(test_hashes)
assert val_hashes.isdisjoint(test_hashes)

for split_name, frame in [
    ("TRAIN", train_df),
    ("VALIDATION", val_df),
    ("TEST", test_df),
]:
    print(f"\n{split_name}: {len(frame)} images")
    print(
        frame["label"]
        .value_counts()
        .sort_index()
        .rename(index={
            0: "negative",
            1: "positive_unlabelled"
        })
    )

for frame in [train_df, val_df, test_df]:
    if frame["label"].nunique() != 2:
        raise ValueError(
            "A split is missing a class. Adjust the subset fraction "
            "or the split settings."
        )

manifest = pd.concat(
    [
        train_df.assign(split="train"),
        val_df.assign(split="validation"),
        test_df.assign(split="test"),
    ],
    ignore_index=True
)

manifest.to_csv(
    OUTPUT_DIR / "dataset_split_manifest.csv",
    index=False
)


# =====================================================================
# 6. BUILD A SAFE PNG INPUT PIPELINE
# =====================================================================
# All input files here were written as PNG by the conversion stage.
# decode_png avoids asking TensorFlow to infer the original format.
# IMPORTANT: each dataset example returns ONE image and ONE label.

def load_image(path, label):
    image_bytes = tf.io.read_file(path)

    image = tf.io.decode_png(
        image_bytes,
        channels=3
    )

    image.set_shape([None, None, 3])

    image = tf.image.resize(
        image,
        [IMAGE_SIZE, IMAGE_SIZE],
        antialias=True
    )

    image = tf.cast(image, tf.float32)
    label = tf.cast(label, tf.float32)

    return image, label


def make_dataset(frame, training=False):
    paths = frame["path"].astype(str).to_numpy()
    labels = frame["label"].astype(np.float32).to_numpy()

    dataset = tf.data.Dataset.from_tensor_slices(
        (paths, labels)
    )

    if training:
        dataset = dataset.shuffle(
            buffer_size=len(frame),
            seed=SEED,
            reshuffle_each_iteration=True
        )

    dataset = dataset.map(
        load_image,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    dataset = dataset.batch(BATCH_SIZE)
    dataset = dataset.prefetch(tf.data.AUTOTUNE)

    return dataset


train_ds = make_dataset(train_df, training=True)
val_ds = make_dataset(val_df)
test_ds = make_dataset(test_df)

print("\nInput pipeline ready.")

# Test one batch before constructing the model.
for sample_images, sample_labels in train_ds.take(1):
    print("Batch image shape:", sample_images.shape)
    print("Batch label shape:", sample_labels.shape)

    assert sample_images.shape[-3:] == (
        IMAGE_SIZE, IMAGE_SIZE, 3
    )
    assert len(sample_images.shape) == 4
    assert len(sample_labels.shape) == 1

print("Input pipeline test passed.")


# =====================================================================
# 7. BUILD THE DENSENET121 + VGG16 FUSION MODEL
# =====================================================================

# Augmentation runs only during training.
augmentation = tf.keras.Sequential(
    [
        tf.keras.layers.RandomFlip("horizontal"),
        tf.keras.layers.RandomRotation(0.05),
        tf.keras.layers.RandomZoom(0.10),
    ],
    name="shared_augmentation"
)

# One image input feeds both CNN branches.
image_input = tf.keras.Input(
    shape=(IMAGE_SIZE, IMAGE_SIZE, 3),
    dtype=tf.float32,
    name="image"
)

augmented = augmentation(image_input)

# Each backbone requires its own preprocessing.
densenet_input = tf.keras.layers.Lambda(
    tf.keras.applications.densenet.preprocess_input,
    name="densenet_preprocessing"
)(augmented)

vgg_input = tf.keras.layers.Lambda(
    tf.keras.applications.vgg16.preprocess_input,
    name="vgg16_preprocessing"
)(augmented)

print("\nLoading ImageNet pretrained backbones...")

densenet_base = tf.keras.applications.DenseNet121(
    include_top=False,
    weights="imagenet",
    input_shape=(IMAGE_SIZE, IMAGE_SIZE, 3)
)

vgg_base = tf.keras.applications.VGG16(
    include_top=False,
    weights="imagenet",
    input_shape=(IMAGE_SIZE, IMAGE_SIZE, 3)
)

# Phase 1: freeze pretrained backbones.
densenet_base.trainable = False
vgg_base.trainable = False

densenet_map = densenet_base(
    densenet_input,
    training=False
)

vgg_map = vgg_base(
    vgg_input,
    training=False
)

densenet_features = tf.keras.layers.GlobalAveragePooling2D(
    name="densenet_global_pool"
)(densenet_map)

vgg_features = tf.keras.layers.GlobalAveragePooling2D(
    name="vgg_global_pool"
)(vgg_map)

densenet_features = tf.keras.layers.Dropout(
    0.25,
    name="densenet_dropout"
)(densenet_features)

vgg_features = tf.keras.layers.Dropout(
    0.25,
    name="vgg_dropout"
)(vgg_features)

fused = tf.keras.layers.Concatenate(
    name="feature_fusion"
)([densenet_features, vgg_features])

fused = tf.keras.layers.BatchNormalization(
    name="fusion_batch_norm"
)(fused)

fused = tf.keras.layers.Dense(
    256,
    activation="relu",
    name="fusion_dense"
)(fused)

fused = tf.keras.layers.Dropout(
    0.40,
    name="fusion_dropout"
)(fused)

output = tf.keras.layers.Dense(
    1,
    activation="sigmoid",
    name="positive_probability"
)(fused)

model = tf.keras.Model(
    inputs=image_input,
    outputs=output,
    name="DenseNet121_VGG16_Fusion"
)

model.summary()


# =====================================================================
# 8. CLASS WEIGHTS
# =====================================================================

train_labels = train_df["label"].to_numpy().astype(int)

classes = np.unique(train_labels)

class_weight_values = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=train_labels
)

class_weights = {
    int(class_id): float(weight)
    for class_id, weight in zip(classes, class_weight_values)
}

print("\nClass weights:", class_weights)


# =====================================================================
# 9. TRAINING CALLBACKS
# =====================================================================

phase1_weights = OUTPUT_DIR / "phase1_best.weights.h5"
phase2_weights = OUTPUT_DIR / "phase2_best.weights.h5"


def compile_model(learning_rate):
    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=learning_rate
        ),
        loss="binary_crossentropy",
        metrics=[
            tf.keras.metrics.BinaryAccuracy(name="accuracy"),
            tf.keras.metrics.AUC(name="auc"),
            tf.keras.metrics.Precision(name="precision"),
            tf.keras.metrics.Recall(name="recall"),
        ]
    )


def make_callbacks(weights_path, patience=4):
    return [
        tf.keras.callbacks.ModelCheckpoint(
            filepath=str(weights_path),
            monitor="val_auc",
            mode="max",
            save_best_only=True,
            save_weights_only=True,
            verbose=1
        ),
        tf.keras.callbacks.EarlyStopping(
            monitor="val_auc",
            mode="max",
            patience=patience,
            restore_best_weights=True,
            verbose=1
        ),
        tf.keras.callbacks.ReduceLROnPlateau(
            monitor="val_loss",
            factor=0.3,
            patience=2,
            min_lr=1e-7,
            verbose=1
        ),
    ]


# =====================================================================
# 10. PHASE 1: TRAIN THE FUSION CLASSIFIER
# =====================================================================

print("\n" + "=" * 65)
print("PHASE 1: TRAINING CLASSIFICATION HEAD")
print("=" * 65)

compile_model(learning_rate=1e-3)

history1 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=HEAD_EPOCHS,
    class_weight=class_weights,
    callbacks=make_callbacks(phase1_weights),
    verbose=1
)

if not phase1_weights.exists():
    raise RuntimeError(
        "Phase 1 did not save a checkpoint. Check the training logs."
    )

model.load_weights(phase1_weights)

best_phase1_auc = max(
    history1.history.get("val_auc", [0.0])
)

print("Best phase-1 validation AUC:", best_phase1_auc)


# =====================================================================
# 11. PHASE 2: FINE-TUNE LATER BACKBONE LAYERS
# =====================================================================

print("\n" + "=" * 65)
print("PHASE 2: FINE-TUNING SELECTED BACKBONE LAYERS")
print("=" * 65)

densenet_base.trainable = True
vgg_base.trainable = True

# Unfreeze only the later layers.
for layer in densenet_base.layers[:-30]:
    layer.trainable = False

for layer in vgg_base.layers[:-20]:
    layer.trainable = False

# Keep batch-normalization layers frozen.
for layer in densenet_base.layers:
    if isinstance(layer, tf.keras.layers.BatchNormalization):
        layer.trainable = False

for layer in vgg_base.layers:
    if isinstance(layer, tf.keras.layers.BatchNormalization):
        layer.trainable = False

compile_model(learning_rate=1e-5)

history2 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=FINE_TUNE_EPOCHS,
    class_weight=class_weights,
    callbacks=make_callbacks(phase2_weights),
    verbose=1
)

best_phase2_auc = max(
    history2.history.get("val_auc", [0.0])
)

# Compare the best validation AUC from each phase.
# The test set is not used to choose the model.
if (
    phase2_weights.exists()
    and best_phase2_auc >= best_phase1_auc
):
    model.load_weights(phase2_weights)
    selected_phase = "fine-tuned"
else:
    model.load_weights(phase1_weights)
    selected_phase = "frozen-backbone"

print("\nSelected phase:", selected_phase)
print("Phase-1 validation AUC:", best_phase1_auc)
print("Phase-2 validation AUC:", best_phase2_auc)


# =====================================================================
# 12. SAVE THE FINAL MODEL
# =====================================================================

final_model_path = (
    OUTPUT_DIR / "densenet121_vgg16_unlabelled_ensemble.keras"
)

model.save(final_model_path)

print("\nSaved model:", final_model_path)


# =====================================================================
# 13. SAVE TRAINING CURVES
# =====================================================================

def combined_history(metric):
    return (
        history1.history.get(metric, [])
        + history2.history.get(metric, [])
    )


for metric in ["loss", "accuracy", "auc"]:
    train_values = combined_history(metric)
    val_values = combined_history("val_" + metric)

    if not train_values or not val_values:
        continue

    plt.figure(figsize=(7, 4))
    plt.plot(train_values, label="Training")
    plt.plot(val_values, label="Validation")
    plt.xlabel("Epoch")
    plt.ylabel(metric)
    plt.title(f"Training and validation {metric}")
    plt.legend()
    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR / f"training_{metric}.png",
        dpi=150
    )
    plt.show()
    plt.close()


# =====================================================================
# 14. FINAL TEST EVALUATION
# =====================================================================
# Evaluate the held-out test set only after model selection is complete.

print("\n" + "=" * 65)
print("FINAL HELD-OUT TEST EVALUATION")
print("=" * 65)

test_probabilities = model.predict(
    test_ds,
    verbose=1
).ravel()

y_test = test_df["label"].to_numpy().astype(int)
y_pred = (test_probabilities >= 0.5).astype(int)

test_accuracy = accuracy_score(y_test, y_pred)

test_precision = precision_score(
    y_test, y_pred, zero_division=0
)

test_recall = recall_score(
    y_test, y_pred, zero_division=0
)

test_f1 = f1_score(
    y_test, y_pred, zero_division=0
)

test_auc = roc_auc_score(
    y_test, test_probabilities
)

print(f"\nAccuracy:  {test_accuracy:.4f}")
print(f"Precision: {test_precision:.4f}")
print(f"Recall:    {test_recall:.4f}")
print(f"F1-score:  {test_f1:.4f}")
print(f"ROC-AUC:   {test_auc:.4f}")

print("\nClassification report:")
print(
    classification_report(
        y_test,
        y_pred,
        labels=[0, 1],
        target_names=["Negative", "Positive unlabelled"],
        zero_division=0
    )
)


# =====================================================================
# 15. CONFUSION MATRIX
# =====================================================================

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=[0, 1]
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Negative", "Positive unlabelled"]
)

disp.plot(values_format="d")
plt.title("Held-out Test Confusion Matrix")
plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "test_confusion_matrix.png",
    dpi=150
)

plt.show()
plt.close()


# =====================================================================
# 16. ROC CURVE
# =====================================================================

fpr, tpr, thresholds = roc_curve(
    y_test,
    test_probabilities
)

plt.figure(figsize=(6, 5))

plt.plot(
    fpr,
    tpr,
    label=f"ROC-AUC = {test_auc:.3f}"
)

plt.plot([0, 1], [0, 1], linestyle="--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Held-out Test ROC Curve")
plt.legend()
plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "test_roc_curve.png",
    dpi=150
)

plt.show()
plt.close()


# =====================================================================
# 17. SAVE METRICS AND PREDICTIONS
# =====================================================================

metrics = {
    "model": "DenseNet121_VGG16_Feature_Fusion",
    "selected_phase": selected_phase,
    "positive_fraction": POSITIVE_FRACTION,
    "image_size": IMAGE_SIZE,
    "batch_size": BATCH_SIZE,
    "train_images": len(train_df),
    "validation_images": len(val_df),
    "test_images": len(test_df),
    "test_accuracy": test_accuracy,
    "test_precision": test_precision,
    "test_recall": test_recall,
    "test_f1": test_f1,
    "test_roc_auc": test_auc,
    "positive_labelled_images_used": 0,
}

pd.DataFrame([metrics]).to_csv(
    OUTPUT_DIR / "test_metrics.csv",
    index=False
)

predictions_df = test_df[
    ["path", "source_path", "label", "hash"]
].copy()

predictions_df["predicted_probability"] = test_probabilities
predictions_df["predicted_label"] = y_pred

predictions_df.to_csv(
    OUTPUT_DIR / "test_predictions.csv",
    index=False
)


# =====================================================================
# 18. FINISHED
# =====================================================================

print("\n" + "=" * 65)
print("TRAINING AND EVALUATION FINISHED")
print("=" * 65)

print("\nClass mapping:")
print("  negative            -> 0")
print("  positive_unlabelled -> 1")
print("  positive_labelled   -> EXCLUDED")

print("\nResults folder:")
print(OUTPUT_DIR)

print("\nFinal metrics:")
for key, value in metrics.items():
    print(f"{key}: {value}")

print("\nSaved output files:")
for item in sorted(OUTPUT_DIR.iterdir()):
    print(" -", item.name)

print("\nDone.")
```
