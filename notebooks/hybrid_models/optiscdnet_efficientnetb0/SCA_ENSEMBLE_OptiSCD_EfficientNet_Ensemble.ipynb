```python

# ================================================================
# EFFICIENTNETB0 + OPTISCDNET-STYLE HYBRID ENSEMBLE
# Dataset: existing Google Drive folders
# ================================================================

import os
import json
import hashlib
import random
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt

from PIL import Image
from google.colab import drive
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, average_precision_score, confusion_matrix,
    classification_report, roc_curve
)

# -------------------- 1. CONFIGURATION --------------------

SEED = 42
IMAGE_SIZE = 224
BATCH_SIZE = 16
INITIAL_EPOCHS = 10
FINE_TUNE_EPOCHS = 20

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# -------------------- 2. MOUNT GOOGLE DRIVE --------------------

if not os.path.isdir("/content/drive/MyDrive"):
    drive.mount("/content/drive")
else:
    print("Google Drive is already mounted.")

DATASET_DIR = "/content/drive/MyDrive/Dataset_SickleCell"

OUTPUT_DIR = os.path.join(
    DATASET_DIR, "EfficientNetB0_OptiSCDNet_Hybrid_Results"
)

if not os.path.isdir(DATASET_DIR):
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_DIR}. Check your Drive location."
    )

# Confirm all dataset folders exist before training.
CLASS_FOLDERS = {
    "negative": 0,
    "positive_unlabelled": 1,
    "positive_labelled": 1
}

for folder in CLASS_FOLDERS:
    folder_path = os.path.join(DATASET_DIR, folder)
    if not os.path.isdir(folder_path):
        raise FileNotFoundError(
            f"Required folder not found: {folder_path}"
        )

os.makedirs(OUTPUT_DIR, exist_ok=True)

MODEL_PATH = os.path.join(
    OUTPUT_DIR, "efficientnetb0_optiscdnet_hybrid.keras"
)
BEST_MODEL_PATH = os.path.join(OUTPUT_DIR, "best_hybrid.keras")

print("TensorFlow:", tf.__version__)
print("Dataset:", DATASET_DIR)
print("Results:", OUTPUT_DIR)

# -------------------- 3. READ THE DATASET --------------------

records = []
extensions = (".jpg", ".jpeg", ".png", ".bmp")

for folder, label in CLASS_FOLDERS.items():
    folder_path = os.path.join(DATASET_DIR, folder)
    count = 0

    for root, _, files in os.walk(folder_path):
        for filename in files:
            if not filename.lower().endswith(extensions):
                continue

            path = os.path.join(root, filename)

            try:
                with Image.open(path) as im:
                    im.verify()

                # Group exact duplicate files to avoid cross-split leakage.
                hasher = hashlib.sha256()
                with open(path, "rb") as file:
                    for chunk in iter(
                        lambda: file.read(1024 * 1024), b""
                    ):
                        hasher.update(chunk)

                records.append({
                    "path": path,
                    "label": label,
                    "folder": folder,
                    "group": hasher.hexdigest()
                })
                count += 1

            except Exception as exc:
                print("Skipping unreadable image:", path, exc)

    print(f"{folder}: {count} readable images")

df = pd.DataFrame(records)

if df.empty:
    raise ValueError("No readable images were found.")

# A byte-identical image must not have conflicting class labels.
conflicts = df.groupby("group")["label"].nunique()
if (conflicts > 1).any():
    raise ValueError(
        "Identical files have conflicting labels. Please inspect the dataset."
    )

print("\nTotal images:", len(df))
print("Class counts:")
print(df["label"].value_counts().sort_index())

# -------------------- 4. TRAIN / VALIDATION / TEST SPLIT --------------------

groups_df = df[["group", "label"]].drop_duplicates("group")

if groups_df["label"].value_counts().min() < 3:
    raise ValueError("Too few unique images in one class for splitting.")

trainval_groups, test_groups = train_test_split(
    groups_df,
    test_size=0.15,
    random_state=SEED,
    stratify=groups_df["label"]
)

train_groups, val_groups = train_test_split(
    trainval_groups,
    test_size=0.1765,
    random_state=SEED,
    stratify=trainval_groups["label"]
)

train_ids = set(train_groups["group"])
val_ids = set(val_groups["group"])
test_ids = set(test_groups["group"])

assert train_ids.isdisjoint(val_ids)
assert train_ids.isdisjoint(test_ids)
assert val_ids.isdisjoint(test_ids)

train_df = df[df["group"].isin(train_ids)].reset_index(drop=True)
val_df = df[df["group"].isin(val_ids)].reset_index(drop=True)
test_df = df[df["group"].isin(test_ids)].reset_index(drop=True)

for name, frame in [
    ("Train", train_df),
    ("Validation", val_df),
    ("Test", test_df)
]:
    if frame["label"].nunique() != 2:
        raise ValueError(f"{name} split does not contain both classes.")

    print(
        f"{name}: {len(frame)} images, "
        f"{frame['group'].nunique()} unique image groups, "
        f"class counts {frame['label'].value_counts().to_dict()}"
    )

for name, frame in [
    ("train", train_df),
    ("validation", val_df),
    ("test", test_df)
]:
    frame.to_csv(
        os.path.join(OUTPUT_DIR, f"{name}_manifest.csv"),
        index=False
    )

# -------------------- 5. IMAGE INPUT PIPELINE --------------------

AUTOTUNE = tf.data.AUTOTUNE

def decode_and_resize(path, label):
    image_bytes = tf.io.read_file(path)
    image = tf.io.decode_image(
        image_bytes, channels=3, expand_animations=False
    )
    image.set_shape([None, None, 3])
    image = tf.image.resize(image, [IMAGE_SIZE, IMAGE_SIZE])

    # Keep 0-255 input for the standard Keras EfficientNetB0.
    image = tf.cast(image, tf.float32)
    label = tf.cast(label, tf.float32)
    return image, label

def make_dataset(frame, training=False):
    paths = frame["path"].astype(str).to_numpy()
    labels = frame["label"].astype(np.float32).to_numpy()

    dataset = tf.data.Dataset.from_tensor_slices((paths, labels))

    if training:
        dataset = dataset.shuffle(
            len(frame), seed=SEED, reshuffle_each_iteration=True
        )

    dataset = dataset.map(
        decode_and_resize, num_parallel_calls=AUTOTUNE
    )
    return dataset.batch(BATCH_SIZE).prefetch(AUTOTUNE)

train_ds = make_dataset(train_df, training=True)
val_ds = make_dataset(val_df)
test_ds = make_dataset(test_df)

# -------------------- 6. OPTISCDNET-STYLE RESIDUAL BLOCK --------------------

def optiscd_block(x, filters, stride=1, name="block"):
    shortcut = x

    x = tf.keras.layers.Conv2D(
        filters, 3, strides=stride, padding="same",
        use_bias=False, name=f"{name}_conv"
    )(x)
    x = tf.keras.layers.BatchNormalization(name=f"{name}_bn1")(x)
    x = tf.keras.layers.Activation("swish", name=f"{name}_act1")(x)

    x = tf.keras.layers.SeparableConv2D(
        filters, 3, padding="same",
        use_bias=False, name=f"{name}_sepconv"
    )(x)
    x = tf.keras.layers.BatchNormalization(name=f"{name}_bn2")(x)

    # Squeeze-and-excitation channel attention.
    se = tf.keras.layers.GlobalAveragePooling2D(
        name=f"{name}_se_pool"
    )(x)
    se = tf.keras.layers.Dense(
        max(filters // 8, 8), activation="swish",
        name=f"{name}_se_fc1"
    )(se)
    se = tf.keras.layers.Dense(
        filters, activation="sigmoid", name=f"{name}_se_fc2"
    )(se)
    se = tf.keras.layers.Reshape((1, 1, filters))(se)
    x = tf.keras.layers.Multiply()([x, se])

    if stride != 1 or int(shortcut.shape[-1]) != filters:
        shortcut = tf.keras.layers.Conv2D(
            filters, 1, strides=stride, padding="same",
            use_bias=False, name=f"{name}_shortcut"
        )(shortcut)
        shortcut = tf.keras.layers.BatchNormalization()(shortcut)

    x = tf.keras.layers.Add()([x, shortcut])
    return tf.keras.layers.Activation("swish")(x)

# -------------------- 7. BUILD THE HYBRID ENSEMBLE --------------------

inputs = tf.keras.Input(
    shape=(IMAGE_SIZE, IMAGE_SIZE, 3),
    dtype=tf.float32,
    name="rgb_image"
)

aug = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal"),
    tf.keras.layers.RandomRotation(0.04),
    tf.keras.layers.RandomZoom(0.08)
], name="augmentation")(inputs)

# Branch A: ImageNet-pretrained EfficientNetB0.
# Do not pass include_preprocessing: it is unsupported in your API.
efficientnet_base = tf.keras.applications.EfficientNetB0(
    include_top=False,
    weights="imagenet",
    input_shape=(IMAGE_SIZE, IMAGE_SIZE, 3)
)
efficientnet_base.trainable = False

eff = efficientnet_base(aug)
eff = tf.keras.layers.GlobalAveragePooling2D()(eff)
eff = tf.keras.layers.Dropout(0.25)(eff)

# Branch B: custom OptiSCDNet-style feature extractor.
custom = tf.keras.layers.Rescaling(1.0 / 255.0)(aug)

custom = tf.keras.layers.Conv2D(
    32, 3, padding="same", use_bias=False
)(custom)
custom = tf.keras.layers.BatchNormalization()(custom)
custom = tf.keras.layers.Activation("swish")(custom)
custom = tf.keras.layers.MaxPooling2D(2)(custom)

custom = optiscd_block(custom, 32, 1, "opti_block1")
custom = optiscd_block(custom, 64, 2, "opti_block2")
custom = optiscd_block(custom, 128, 2, "opti_block3")
custom = optiscd_block(custom, 192, 2, "opti_block4")

custom = tf.keras.layers.GlobalAveragePooling2D()(custom)
custom = tf.keras.layers.Dense(256, activation="swish")(custom)
custom = tf.keras.layers.Dropout(0.30)(custom)

# Feature fusion: both branches contribute to one classifier.
fused = tf.keras.layers.Concatenate(name="feature_fusion")(
    [eff, custom]
)
fused = tf.keras.layers.BatchNormalization()(fused)
fused = tf.keras.layers.Dense(256, activation="swish")(fused)
fused = tf.keras.layers.Dropout(0.40)(fused)
fused = tf.keras.layers.Dense(64, activation="swish")(fused)

outputs = tf.keras.layers.Dense(
    1, activation="sigmoid", name="positive_probability"
)(fused)

model = tf.keras.Model(
    inputs, outputs, name="EfficientNetB0_OptiSCDNet_Hybrid"
)

model.summary()

# -------------------- 8. TRAINING CONFIGURATION --------------------

counts = train_df["label"].value_counts().to_dict()
n_train = len(train_df)

class_weight = {
    0: n_train / (2.0 * counts[0]),
    1: n_train / (2.0 * counts[1])
}
print("Class weights:", class_weight)

def compile_model(lr):
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=lr),
        loss="binary_crossentropy",
        metrics=[
            tf.keras.metrics.BinaryAccuracy(name="accuracy"),
            tf.keras.metrics.Precision(name="precision"),
            tf.keras.metrics.Recall(name="recall"),
            tf.keras.metrics.AUC(name="auc")
        ]
    )

def make_callbacks(phase):
    return [
        tf.keras.callbacks.ModelCheckpoint(
            BEST_MODEL_PATH,
            monitor="val_auc",
            mode="max",
            save_best_only=True,
            verbose=1
        ),
        tf.keras.callbacks.EarlyStopping(
            monitor="val_auc",
            mode="max",
            patience=5,
            restore_best_weights=True,
            verbose=1
        ),
        tf.keras.callbacks.ReduceLROnPlateau(
            monitor="val_loss",
            factor=0.3,
            patience=2,
            min_lr=1e-7,
            verbose=1
        ),
        tf.keras.callbacks.CSVLogger(
            os.path.join(OUTPUT_DIR, f"training_{phase}.csv")
        )
    ]

# -------------------- 9. PHASE 1: TRAIN NEW BRANCHES AND HEAD --------------------

compile_model(1e-3)

print("\nPHASE 1: EfficientNetB0 frozen.")
history1 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=INITIAL_EPOCHS,
    class_weight=class_weight,
    callbacks=make_callbacks("phase1")
)

# -------------------- 10. PHASE 2: FINE-TUNE EFFICIENTNET --------------------

print("\nPHASE 2: Fine-tuning upper EfficientNetB0 layers.")

efficientnet_base.trainable = True
fine_tune_from = max(0, len(efficientnet_base.layers) - 40)

for i, layer in enumerate(efficientnet_base.layers):
    layer.trainable = (
        i >= fine_tune_from
        and not isinstance(layer, tf.keras.layers.BatchNormalization)
    )

print(
    "Trainable EfficientNet layers:",
    sum(int(layer.trainable) for layer in efficientnet_base.layers),
    "of", len(efficientnet_base.layers)
)

compile_model(1e-5)

history2 = model.fit(
    train_ds,
    validation_data=val_ds,
    initial_epoch=len(history1.epoch),
    epochs=len(history1.epoch) + FINE_TUNE_EPOCHS,
    class_weight=class_weight,
    callbacks=make_callbacks("phase2")
)

# Load the checkpoint with the best validation AUC.
if os.path.isfile(BEST_MODEL_PATH):
    model = tf.keras.models.load_model(BEST_MODEL_PATH, compile=False)

model.save(MODEL_PATH)
print("\nSaved model:", MODEL_PATH)

# -------------------- 11. HELD-OUT TEST EVALUATION --------------------

y_true = test_df["label"].to_numpy(dtype=int)
y_scores = model.predict(test_ds, verbose=1).reshape(-1)
y_pred = (y_scores >= 0.5).astype(int)

tn, fp, fn, tp = confusion_matrix(
    y_true, y_pred, labels=[0, 1]
).ravel()

metrics = {
    "test_images": int(len(y_true)),
    "threshold": 0.5,
    "accuracy": float(accuracy_score(y_true, y_pred)),
    "precision": float(precision_score(y_true, y_pred, zero_division=0)),
    "sensitivity_recall": float(recall_score(y_true, y_pred, zero_division=0)),
    "specificity": float(tn / (tn + fp)) if (tn + fp) else 0.0,
    "f1_score": float(f1_score(y_true, y_pred, zero_division=0)),
    "roc_auc": float(roc_auc_score(y_true, y_scores)),
    "average_precision": float(average_precision_score(y_true, y_scores)),
    "tn": int(tn), "fp": int(fp), "fn": int(fn), "tp": int(tp)
}

print("\nHELD-OUT TEST RESULTS")
print(pd.Series(metrics).to_string())
print("\nClassification report:")
print(classification_report(
    y_true, y_pred,
    labels=[0, 1],
    target_names=["Negative", "Positive"],
    zero_division=0
))

with open(os.path.join(OUTPUT_DIR, "test_metrics.json"), "w") as f:
    json.dump(metrics, f, indent=2)

pred_df = test_df.copy()
pred_df["true_label"] = y_true
pred_df["positive_probability"] = y_scores
pred_df["predicted_label"] = y_pred
pred_df.to_csv(
    os.path.join(OUTPUT_DIR, "test_predictions.csv"), index=False
)

# -------------------- 12. FOLDER SUBGROUP RESULTS --------------------

subgroup_rows = []

for folder_name in CLASS_FOLDERS:
    subset = pred_df[pred_df["folder"] == folder_name]
    if subset.empty:
        continue

    yt = subset["true_label"].to_numpy(dtype=int)
    yp = subset["predicted_label"].to_numpy(dtype=int)
    ys = subset["positive_probability"].to_numpy(dtype=float)

    subgroup_rows.append({
        "folder": folder_name,
        "images": len(subset),
        "accuracy": accuracy_score(yt, yp),
        "precision": precision_score(yt, yp, zero_division=0),
        "recall": recall_score(yt, yp, zero_division=0),
        "f1": f1_score(yt, yp, zero_division=0),
        "roc_auc": (
            roc_auc_score(yt, ys) if len(np.unique(yt)) == 2 else None
        )
    })

subgroup_df = pd.DataFrame(subgroup_rows)
subgroup_df.to_csv(
    os.path.join(OUTPUT_DIR, "folder_subgroup_metrics.csv"), index=False
)
print("\nFolder subgroup results:")
print(subgroup_df.to_string(index=False))

# -------------------- 13. CONFUSION MATRIX --------------------

cm = confusion_matrix(y_true, y_pred, labels=[0, 1])

plt.figure(figsize=(6, 5))
plt.imshow(cm, interpolation="nearest")
plt.title("Hybrid Test Confusion Matrix")
plt.colorbar()
plt.xticks([0, 1], ["Predicted Negative", "Predicted Positive"])
plt.yticks([0, 1], ["Actual Negative", "Actual Positive"])

for i in range(2):
    for j in range(2):
        plt.text(j, i, str(cm[i, j]), ha="center", va="center", fontsize=14)

plt.xlabel("Predicted label")
plt.ylabel("True label")
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "confusion_matrix.png"), dpi=300)
plt.show()
plt.close()

# -------------------- 14. ROC CURVE --------------------

fpr, tpr, _ = roc_curve(y_true, y_scores)
auc_value = roc_auc_score(y_true, y_scores)

plt.figure(figsize=(7, 6))
plt.plot(fpr, tpr, label=f"Hybrid ROC (AUC = {auc_value:.4f})")
plt.plot([0, 1], [0, 1], linestyle="--", label="Chance")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("EfficientNetB0 + OptiSCDNet-Style Hybrid ROC")
plt.legend(loc="lower right")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "roc_curve.png"), dpi=300)
plt.show()
plt.close()

# -------------------- 15. SAVE CONFIGURATION --------------------

config = {
    "dataset": DATASET_DIR,
    "output_dir": OUTPUT_DIR,
    "image_size": IMAGE_SIZE,
    "batch_size": BATCH_SIZE,
    "seed": SEED,
    "class_mapping": CLASS_FOLDERS,
    "efficientnet": "ImageNet-pretrained EfficientNetB0",
    "custom_branch": "OptiSCDNet-style residual separable CNN",
    "fusion": "Concatenated branch features and one sigmoid classifier",
    "split": "Exact SHA-256 duplicate groups kept within one split",
    "caveats": [
        "This custom branch is inspired by an OptiSCDNet-style design; it is not a verified reproduction of a separate original model.",
        "Hash grouping does not detect visually identical files encoded differently.",
        "Patient-level leakage cannot be ruled out without patient identifiers.",
        "Black annotation rectangles in positive_labelled images may create shortcut learning."
    ]
}

with open(os.path.join(OUTPUT_DIR, "run_config.json"), "w") as f:
    json.dump(config, f, indent=2)

print("\n" + "=" * 65)
print("HYBRID TRAINING AND EVALUATION COMPLETE")
print("=" * 65)
print("Final model:", MODEL_PATH)
print("Best checkpoint:", BEST_MODEL_PATH)
print("Metrics:", os.path.join(OUTPUT_DIR, "test_metrics.json"))
print("Predictions:", os.path.join(OUTPUT_DIR, "test_predictions.csv"))
print("Subgroup metrics:", os.path.join(OUTPUT_DIR, "folder_subgroup_metrics.csv"))
print("All results:", OUTPUT_DIR)
```
